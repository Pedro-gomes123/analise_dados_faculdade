```python
import requests
import pandas as pd

# source venv/bin/activate
```

```python

URL = (
    "https://wis2bra.inmet.gov.br/oapi/"
    "collections/urn:wmo:md:br-inmet:synop/items"
)


def coletar_inmet():

    params = {
        "limit": 100
    }

    response = requests.get(
        URL,
        params=params,
        timeout=30
    )

    print("URL:", response.url)
    print("STATUS:", response.status_code)

    response.raise_for_status()

    dados = response.json()

    print("Tipo:", dados.get("type"))

    features = dados.get("features", [])

    print("Registros encontrados:", len(features))

    if not features:
        return pd.DataFrame()

    registros = []

    for feature in features:

        propriedades = feature.get(
            "properties",
            {}
        )

        registros.append(propriedades)

    return pd.DataFrame(registros)


df = coletar_inmet()

print("\nCOLUNAS:")
print(df.columns.tolist())

print("\nDADOS:")
print(df.head())
```

```python
BASE_URL = (
    "https://geoportal.apac.pe.gov.br/server/rest/services/"
    "met_monitoramento_chuvas_pe/MapServer/6/query"
)

params = {
    "where": "1=1",
    "outFields": "*",
    "returnGeometry": "false",
    "f": "json"
}

response = requests.get(
    BASE_URL,
    params=params,
    timeout=60
)

print("STATUS:", response.status_code)
print("URL:", response.url)
print("CONTENT-TYPE:", response.headers.get("Content-Type"))
print("\nRESPOSTA:")
print(response.text[:2000])
```
