```python
import requests as rq
import pandas as pd

# source venv/bin/activate
```

```python

URL = (
    "https://wis2bra.inmet.gov.br/oapi/"
    "collections/urn:wmo:md:br-inmet:synop/items"
)


def coletar_inmet():

    params = {
        "limit": 100
    }

    response = requests.get(
        URL,
        params=params,
        timeout=30
    )

    print("URL:", response.url)
    print("STATUS:", response.status_code)

    response.raise_for_status()

    dados = response.json()

    print("Tipo:", dados.get("type"))

    features = dados.get("features", [])

    print("Registros encontrados:", len(features))

    if not features:
        return pd.DataFrame()

    registros = []

    for feature in features:

        propriedades = feature.get(
            "properties",
            {}
        )

        registros.append(propriedades)

    return pd.DataFrame(registros)


df = coletar_inmet()

print("\nCOLUNAS:")
print(df.columns.tolist())

print("\nDADOS:")
print(df.head())
```

```python


url = "https://geoportal.apac.pe.gov.br/server/rest/services/met_monitoramento_chuvas_pe/MapServer/6/query"


limite = (pd.Timestamp.now() - pd.Timedelta(minutes=10)).strftime("%Y-%m-%d %H:%M:%S")

params = {
    "where": f"ultima_leitura_data_hora >= '{limite}'",
    "outFields": "*",
    "f": "json"
}

response = rq.get(url, params=params, timeout=60, verify=False)

dados = response.json()    # converte o texto JSON em dicionário Python

print(dados)
```
